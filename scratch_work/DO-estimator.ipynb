```python
import sys

def calculate_wastewater_oxygen():
    print("--- Wastewater DO & OTR Calculator ---")
    
    # --- 1. Constants & Assumptions ---
    # Basin Dimensions
    WIDTH_M = 10.0
    LENGTH_M = 10.0
    
    # Physics Constants at 20°C
    # Standard saturation of clean water at 20°C (mg/L or g/m^3)
    CS_20_CLEAN = 9.09 
    
    # Wastewater Factors (Averages for Domestic Wastewater)
    # Alpha (α): Affects the RATE of transfer (interference by detergents/soaps)
    # Range is typically 0.4 - 0.8 for raw wastewater. We use 0.65 as an average.
    ALPHA_FACTOR = 0.65 
    
    # Beta (β): Affects the SATURATION limit (interference by solids/salinity)
    # Range is typically 0.95 - 0.98. We use 0.95.
    BETA_FACTOR = 0.95
    
    # --- 2. User Input ---
    try:
        height_input = input("Enter the water depth (height) in meters: ")
        height_m = float(height_input)
        
        if height_m <= 0:
            print("Error: Height must be a positive number.")
            return
    except ValueError:
        print("Error: Please enter a valid numeric value.")
        return

    # --- 3. Calculations ---
    
    # Volume in cubic meters (m^3)
    volume_m3 = LENGTH_M * WIDTH_M * height_m
    
    # Volume in Liters (for visualization, though 1 m^3 = 1000 L)
    volume_liters = volume_m3 * 1000

    # A. Calculate Saturation Concentration (The "Holding Capacity")
    # Formula: Clean Water Saturation * Beta Factor
    sat_conc_wastewater = CS_20_CLEAN * BETA_FACTOR
    
    # B. Calculate Total Dissolved Oxygen Mass at Saturation
    # Formula: Volume (m^3) * Concentration (g/m^3) = Mass (grams)
    total_do_mass_grams = volume_m3 * sat_conc_wastewater
    total_do_mass_kg = total_do_mass_grams / 1000.0

    # C. OTR Adjustment (Information for Aeration Sizing)
    # This shows how much "Standard" oxygen you need to buy to get 1 unit of "Actual" oxygen
    transfer_efficiency_penalty = (1 / ALPHA_FACTOR) * 100

    # --- 4. Output ---
    print("\n" + "="*40)
    print(f"RESULTS FOR {LENGTH_M}m x {WIDTH_M}m x {height_m}m BASIN")
    print("="*40)
    
    print(f"\n1. VOLUMETRICS")
    print(f"   Total Volume:       {volume_m3:,.2f} m³")
    print(f"   Total Liters:       {volume_liters:,.0f} L")

    print(f"\n2. TOTAL DISSOLVED OXYGEN (CAPACITY)")
    print(f"   (Max oxygen this volume can hold at 20°C w/ Beta {BETA_FACTOR})")
    print(f"   Saturation Conc:    {sat_conc_wastewater:.2f} mg/L")
    print(f"   Total Oxygen Mass:  {total_do_mass_kg:.3f} kg")
    
    print(f"\n3. AERATION / OTR FACTORS")
    print(f"   Alpha (α) Factor:   {ALPHA_FACTOR} (Assumed for Domestic Wastewater)")
    print(f"   Impact: Your aerators will transfer oxygen at {ALPHA_FACTOR:.0%} ")
    print(f"   efficiency compared to clean water.")
    print("="*40)

if __name__ == "__main__":
    calculate_wastewater_oxygen()
```
